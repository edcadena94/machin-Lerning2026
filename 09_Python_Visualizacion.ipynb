## 2 .plot()

Multiples parametros

plt.plot(x, y, style, label, linewidth=1, alpha=1)

- style: cadena de texto con formato (color, tipo...) [color][marker][line]
- label: referencia para leyenda
- linewidth: brocha
- alpha: transparencia

Otros: https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html

```python
y_data = np.random.randint(100,size=(10))
x_data = np.arange(10)

fig,axis = plt.subplots(1,1)
axis.plot(x_data,y_data,'gh--',label='dollars',linewidth=1,alpha=0.6) # creacion del grafico
axis.legend(loc='best') # representar la leyenda
plt.show()
```

Colores disponibles: b (blue), g (green), r (red), c (cyan), m (magenta), y (yellow), k (black), w (white)

Marcadores disponibles: . (point), , (pixel), o (circle), v (triangle_down), ^ (triangle_up), < (triangle_left), > (triangle_right), 1/2/3/4 (tri marks), s (square), p (pentagon), * (star), h (hexagon1), H (hexagon2), + (plus), x (x marker), D (diamond), d (thin_diamond), _ (hline)

Estilos de linea: - (solid), -- (dashed), -. (dash-dot), : (dotted)

## 4 Tipos de graficos

Ejemplos disponibles como referencia: https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.subplots.html

```python
# Horizontal bar
fig, ax = plt.subplots()
# Example data
people = ('PhD', 'MSc', 'BSc', 'N/A')
y_pos = np.arange(len(people))
percentage = [5,45,25,25]

# print(y_pos)
ax.barh(y_pos, percentage, align='center',
        color='green', edgecolor='red')
ax.set_yticks(y_pos)
ax.set_yticklabels(people)
#ax.invert_yaxis()  # labels read top-to-bottom
ax.set_xlabel('Percentage of workforce')
ax.set_title('Level of education in the company')
plt.show()
```

```python
ls_count = (284487, 244560, 208493, 196764)
cities = ('Madrid', 'Barcelona', 'Sevilla', 'Valencia')
data = np.array(ls_count)/1000
y_pos = np.arange(data.size)

width = 0.35       # the width of the bars
opacity = 0.7

fig, ax = plt.subplots()
rects1 = ax.barh(y_pos, data, width, alpha=opacity)

rects1[3].set_color('#872f29')
rects1[2].set_color('#ffcb59')
rects1[1].set_color('#d6ee39')
rects1[0].set_color('#6ade30')

ax.set_yticks(y_pos)
ax.set_yticklabels(cities)

# add some text for labels, title and axes ticks
ax.set_ylabel('Ciudades')
ax.set_xlabel('Km de carril bici')

ax.legend((rects1[0], rects1[1], rects1[2], rects1[3]), cities)

plt.show()
```

## 6 frame.plot()

- (solo para Series) label: referencia para la leyenda

```python
#scatter de clases, coloreando segun clase en matplotlib
iris = sns.load_dataset('iris')

# preparacion de los datos
data = []
colors = ["red", "green", "blue"]
groups = ["setosa", "versicolor", "virginica"]
columns = ["petal_length", "petal_width"]

for i in groups:
    # busqueda por clase definida
    data.append(iris[iris["species"] == i][columns].values)

#creacion de la grafica
fig = plt.figure()
ax = fig.add_subplot()

for count, d in enumerate(data):
    # separar las variables del grafico
    x, y = d[:,0], d[:,1]
    ax.scatter(x, y, alpha = 0.4, c = colors[count -1], edgecolors = 'none', s = 30, label = groups[count - 1])

#etiquetas
ax.set_xlabel(columns[0])
ax.set_ylabel(columns[1])

plt.legend(loc='best')
plt.title('Matplot scatter plot')
plt.show()
```

## Solucion Ejercicios 8.1 - matplotlib

```python
# Crear lista de anios 2000 a 2020 (eje x) y 4 secuencias aleatorias (eje y)
anios = list(range(2000, 2021))
n = len(anios)

y0 = np.random.randint(1000, 50000, size=n)
y1 = np.random.randint(1000, 50000, size=n)
y2 = np.random.randint(1000, 50000, size=n)
y3 = np.random.randint(1000, 50000, size=n)

nombres = ["Hawaii", "San Marino", "Islas Feroe", "Guayana"]
```

```python
# Representar las 4 secuencias en UNA sola figura, con leyenda
fig, ax = plt.subplots(1, 1, figsize=(10, 5))

ax.plot(anios, y0, 'g-', label=nombres[0])
ax.plot(anios, y1, 'b-', label=nombres[1])
ax.plot(anios, y2, 'r-', label=nombres[2])
ax.plot(anios, y3, 'm-', label=nombres[3])

ax.set_title("Datos aleatorios")
ax.set_xlabel("Anio")
ax.set_ylabel("GDP")
ax.legend(loc='best')
plt.grid()
plt.show()
```

```python
# Representar los mismos datos en 4 subfiguras (axes) distintas, parte de la misma figura
# Con titulo general, leyenda en cada una, nombres de ejes, y una anotacion con flecha

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
fig.suptitle("Datos aleatorios")

datos = [y0, y1, y2, y3]
colores = ['g-', 'b-', 'r-', 'm-']

for i, ax in enumerate(axes.flatten()):
    ax.plot(anios, datos[i], colores[i], label=nombres[i])
    ax.set_xlabel("Anio")
    ax.set_ylabel("GDP")
    ax.legend(loc='best')

# Anotacion (texto + flecha) en el primer grafico: senialar el valor maximo de Hawaii
idx_max = int(np.argmax(y0))
axes[0, 0].annotate('Maximo GDP',
                    xy=(anios[idx_max], y0[idx_max]),
                    xytext=(anios[2], max(y0) * 0.7),
                    arrowprops=dict(facecolor='red', shrink=0.05))

plt.tight_layout()
plt.show()
```

## Solucion Ejercicios 8.2 - Seaborn

Dataset elegido: **mpg** (consumo de combustible de coches) - distinto a tips, iris y flights, y disponible en https://github.com/mwaskom/seaborn-data

Variables numericas continuas: mpg, displacement, horsepower, weight, acceleration. Variable categorica: origin.

```python
# Cargar el dataset y explorarlo brevemente
mpg = sns.load_dataset('mpg')
mpg = mpg.dropna()  # horsepower tiene algunos NaN, los eliminamos para los graficos
display(mpg.sample(5))
mpg.info()
```

```python
# Representar visualmente la relacion de cada pareja de atributos (variables)
sns.pairplot(mpg)
plt.show()
```

```python
# Version con la variable categorica 'origin' como color (opcional, mas informativa)
sns.pairplot(mpg, hue='origin')
plt.show()
```

```python
# Escoger una pareja (ambos valores continuos y numericos) y demostrar su correlacion
# Pareja elegida: weight (peso) y mpg (millas por galon)

# Correlacion numerica
print("Correlacion weight vs mpg:", mpg['weight'].corr(mpg['mpg']))

# Correlacion visual con recta de regresion e intervalo de confianza
sns.regplot(x='weight', y='mpg', data=mpg)
plt.show()
```

```python
# Representar los valores de dos variables (x e y, numericas) en base a otra variable (categorica)
# x = horsepower, y = mpg, categorica = origin
sns.relplot(x='horsepower', y='mpg', hue='origin', data=mpg, kind='scatter')
plt.show()
```

```python
# Distribucion de valores (media, distribucion, outliers) de una variable numerica continua
# Variable elegida: mpg

# Boxplot: muestra mediana, cuartiles y outliers
sns.boxplot(y='mpg', data=mpg)
plt.show()

# Histograma + density plot: muestra la forma de la distribucion
sns.displot(mpg['mpg'], kde=True, bins=15, rug=True)
plt.show()
```

```python
# Comparar la distribucion de esa variable con otra (continua y numerica)
# para representar cuantitativamente la densidad de elementos por valor en ambas variables
# Pareja: mpg vs weight, con density plot bi-variable (kde)
sns.jointplot(x='weight', y='mpg', kind='kde', data=mpg)
plt.show()
```

```python
# Ejemplo de una variable a la que un modelo de regresion lineal de orden > 1
# se ajusta mejor que uno de orden == 1
# La relacion horsepower vs mpg es claramente curva (no lineal)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Orden 1: la recta no captura la curvatura de los datos
sns.regplot(x='horsepower', y='mpg', data=mpg, order=1, ax=axes[0])
axes[0].set_title('Regresion orden 1 (recta) - ajuste pobre')

# Orden 2: la curva sigue mucho mejor la forma de los datos
sns.regplot(x='horsepower', y='mpg', data=mpg, order=2, ax=axes[1])
axes[1].set_title('Regresion orden 2 (curva) - mejor ajuste')

plt.show()
```

**Conclusion del ultimo ejercicio:** la relacion entre horsepower y mpg no es lineal — a mas caballos de fuerza, el consumo empeora rapidamente al principio pero luego se estabiliza. Por eso la regresion de orden 2 (parabola) sigue la nube de puntos mucho mejor que la recta de orden 1, que se queda por encima de los datos en los extremos y por debajo en el centro.
