# 📖 R1 — Parsear strings y crear DataFrame

### 🔍 Como reconocerlo
El enunciado da strings con un formato fijo tipo `"Nombre - Año"` o `"Cosa (dato)"` y pide separar las partes en columnas de un DataFrame. Ejemplo real del demo: peliculas con el año concatenado.

### 🧠 Plan mental (paso a paso)
1. Mira el separador: ¿es `" - "`, `"("`, `","`? ¿Aparece SOLO una vez o puede repetirse?
2. Si el separador puede aparecer dentro del texto (ej. un guion en el titulo), separa desde la DERECHA con `rsplit(sep, 1)`.
3. Crea el DataFrame desde el dict primero, luego transforma columnas con `.apply(lambda ...)`.
4. **SIEMPRE** `.strip()` cada pieza.
5. Verifica el TIPO del año en expected_output: ¿"1999" (str) o 1999 (int)?

### ⚠️ Trampas
- Espacios invisibles → el assert falla aunque "se vea igual". Solucion: strip().
- Titulos con el separador dentro ("El Senyor de los Anillos, La Comunidad...") → usa rsplit desde la derecha si el separador es ambiguo.
- Devolver int cuando esperan str (o al reves).

### 📋 Plantilla

````python
# PLANTILLA R1 - adapta nombre de funcion, separador y columnas
def extract_year(movies):
    df = pd.DataFrame(movies)
    df['Year'] = df['Name'].apply(lambda x: x.rsplit('-', 1)[1].strip())
    df['Name'] = df['Name'].apply(lambda x: x.rsplit('-', 1)[0].strip())
    return df

# --- test ---
movies = {'Name': ["Matrix - 1999", "Cadena Perpetua - 1994", "Parasitos - 2019"]}
expected = pd.DataFrame({'Name': ["Matrix","Cadena Perpetua","Parasitos"],
                         'Year': ["1999","1994","2019"]})
output = extract_year(movies)
display(output)
assert expected.equals(output)
print("OK R1")
````

### 🔄 Variantes que puede poner el profe
- Tres partes en vez de dos ("Artista - Cancion (Año)") → separa en 2 pasos, primero un separador y luego el otro.
- Que el año sea int → agrega `int(...)` al final del lambda.
- Que pida una Serie en vez de DataFrame → `return df['Year']`.

# 📖 R2 — Duplicados (detectar o eliminar)

### 🔍 Como reconocerlo
Palabras: "duplicados", "repetidos", "primera ocurrencia", "sin elementos repetidos", "True/False si esta repetido".

### 🧠 Plan mental — decide cual de las 3 sub-tareas es:
- **A) MARCAR duplicados** (mask booleana, True desde la 2da ocurrencia) → `pd.Series(arr).duplicated()`
- **B) MARCAR primeras ocurrencias** (lo inverso) → `~pd.Series(arr).duplicated()`
- **C) ELIMINAR duplicados** → decide si importa el ORDEN:
  - Orden NO importa / ordenado esta bien → `np.unique(lista)`
  - Preservar orden de aparicion → `list(dict.fromkeys(lista))`
  - En DataFrame → `df.drop_duplicates()`

### ⚠️ Trampas
- `np.unique` ORDENA siempre → si el expected preserva el orden original y tu resultado sale ordenado, el assert falla. Usa `dict.fromkeys`.
- `set()` NO garantiza orden.
- ¿Devuelve array o Series? Si el test compara con `np.array_equal`, agrega `.values` al final.

### 📋 Plantillas (las 3)

````python
# PLANTILLA R2-A: marcar duplicados (True desde la 2da ocurrencia)
def find_duplicates(numbers):
    return pd.Series(numbers).duplicated().values

numbers = np.array([4, 4, 4, 3, 8, 1, 9, 6, 1, 5, 9, 4, 0, 2])
expected = np.array([False,True,True,False,False,False,False,False,True,False,True,True,False,False])
assert np.array_equal(expected, find_duplicates(numbers))
print("OK R2-A")
````

````python
# PLANTILLA R2-C: eliminar duplicados -> DataFrame
def remove_duplicates(lista):
    return pd.DataFrame(np.unique(lista))            # si el expected sale ORDENADO

def remove_duplicates_orden(lista):
    return pd.DataFrame(list(dict.fromkeys(lista)))  # si el expected preserva ORDEN de aparicion

serie = ['a','b','c','a','c','a','g']
expected = pd.DataFrame(['a','b','c','g'])
assert expected.equals(remove_duplicates(serie))
print("OK R2-C")
````

# 📖 R3 — Sistemas de ecuaciones

### 🔍 Como reconocerlo
Ves ecuaciones con x, y (y quiza z). Pide "resolver" y casi siempre "verificar por codigo".

### 🧠 Plan mental
1. Ordena cada ecuacion: variables a la izquierda EN EL MISMO ORDEN, constante a la derecha.
2. Matriz A = coeficientes (fila por ecuacion, columna por variable). Vector b = constantes.
3. `np.linalg.solve(A, b)` y desempaqueta tantas variables como haya.
4. Verifica CADA ecuacion con `math.isclose(expresion, constante)`.

### ⚠️ Trampas
- Coeficiente implicito: "x - 3y" → el coeficiente de x es **1**, no 0. "-y" → **-1**.
- Si una ecuacion viene desordenada ("-7y + 16x = 80"), REORDENALA mentalmente a (16, -7) antes de escribir la fila.
- Usar `==` en vez de `isclose` → falla por decimales flotantes.

### 📋 Plantilla

````python
# PLANTILLA R3 - sistema 2x2 y 3x3
import numpy.linalg as linalg

# A: -9x + 4y = 20 ; 16x - 7y = 80   (ojo: la 2da venia como "-7y + 16x", reordenada)
a = np.array([[-9, 4], [16, -7]])
b = np.array([20, 80])
x, y = linalg.solve(a, b)
print(f'x={x} y={y}')
print(math.isclose(-9*x + 4*y, 20))
print(math.isclose(16*x - 7*y, 80))

# B: x - 2y + 3z = 7 ; 2x + y + z = 4 ; -3x + 2y - 2z = -10
A2 = np.array([[1,-2,3],[2,1,1],[-3,2,-2]])
b2 = np.array([7,4,-10])
x, y, z = linalg.solve(A2, b2)
print(f'x={x} y={y} z={z}')
print(math.isclose(x - 2*y + 3*z, 7))
print(math.isclose(2*x + y + z, 4))
print(math.isclose(-3*x + 2*y - 2*z, -10))
````

# 📖 R4 — Series mixtas + loc vs iloc

### 🔍 Como reconocerlo
Pide crear una Series con "mas de un tipo basico", "indices de tipo string", y "demostrar" que indices posicionales y semanticos dan lo mismo con slicing.

### 🧠 Plan mental
1. Crea la Series mezclando int, float, str, bool con `index=list('abc...')`.
2. **REGLA DE ORO:** `.loc[:'d']` INCLUYE la 'd'. `.iloc[:4]` NO incluye la posicion 4 (toma 0,1,2,3).
3. Para que coincidan: la etiqueta N-esima con loc equivale a iloc[:N] (loc hasta la letra en posicion N-1).
4. "Demostrar" = comparar con `==` e imprimir, o `assert (comparacion).all()`.

### 📋 Plantilla

````python
# PLANTILLA R4
serie = pd.Series([1, 2, 3, 4, 5, True, "acelga", 4.5, 9, 10], index=list('abcdefghij'))

# los primeros 4 elementos: posicional [:4] == semantico [:'d']
print(serie.iloc[:4])
print(serie.loc[:'d'])
comparacion = serie.iloc[:4] == serie.loc[:'d']
print(comparacion)
assert comparacion.all()
print("OK R4")
````

### ⚠️ Trampas
- Si el DataFrame tiene indices ENTEROS desordenados (ej. index=[1,0]), `frame[0]` es ambiguo → SIEMPRE usa loc (busca etiqueta 0) o iloc (busca posicion 0), que dan resultados DISTINTOS.
- El enunciado puede pedir "al menos 4 elementos" → cuenta bien cuantos selecciona tu slice.

# 📖 R5 — DataFrame aleatorio + NaN + conteo

### 🔍 Como reconocerlo
"Crear DataFrame NxM con aleatorios en rango [a,b]", "reemplazar los menores/mayores de X por NaN", "cuantos NaN por fila/columna".

### 🧠 Plan mental (3 pasos que casi siempre van juntos)
1. **Crear:** `pd.DataFrame(np.random.randint(b+1, size=(N,M)))` — ojo: randint EXCLUYE el limite superior, por eso rango [0,20] = randint(21).
2. **Reemplazar:** dos formas equivalentes:
   - `df[df < 5] = np.nan` (directa, facil de recordar)
   - `df.where(df >= 5, np.nan)` — OJO: where MANTIENE donde la condicion es True y reemplaza donde es False (es "al reves" de lo intuitivo).
3. **Contar:** `df.isna().sum(axis=1)` por FILA, `df.isna().sum(axis=0)` o sin axis por COLUMNA.

### ⚠️ Trampas
- randint(20) da [0,19], NO incluye el 20. Para [0,20] usa randint(21) o randint(0,21).
- axis: el eje que DESAPARECE del resultado es el que se recorre. Por fila = axis=1. Por columna = axis=0.
- `isna()` e `isnull()` son EXACTAMENTE lo mismo.

### 📋 Plantilla

````python
# PLANTILLA R5 - los 3 pasos completos
df = pd.DataFrame(np.random.randint(21, size=(10,10)))   # rango [0,20]
display(df)

df[df < 5] = np.nan                    # reemplazar menores de 5 por NaN
display(df)

print("NaN por fila:")
print(df.isna().sum(axis=1))

print("NaN por columna:")
print(df.isna().sum(axis=0))
````

### 🔄 Variantes
- "Reemplazar por 0 en vez de NaN" → `df[df < 5] = 0`
- "Reemplazar los NaN por la media" → `df.fillna(df.mean())`
- "Eliminar filas con NaN" → `df.dropna()` (columnas: `axis=1`)
- Con np.where puro: `np.where(np.isnan(data), 0, data)`

# 📖 R6 — Generadores (yield)

### 🔍 Como reconocerlo
Palabras magicas: "generador", "cada vez que se invoque", "numero infinito de veces", "simular arange", "sin ninguna libreria", "plain Python".

### 🧠 Plan mental
1. Estructura FIJA: `while True:` + `yield valor` + actualizar valor.
2. Se consume con `next(g)`, NO llamando la funcion otra vez.
3. La palabra "infinito" = OBLIGATORIO el while True (sin el, StopIteration a la 2da llamada).
4. **PROHIBIDO usar NumPy** si dice "sin librerias" — aunque funcione, es incorrecto.

### 📋 Plantilla (memoriza LA ESTRUCTURA, cambia solo la operacion)

````python
# PLANTILLA R6 - progresion aritmetica (suma step)
def frange(init, step):
    while True:
        yield init
        init += step

g = frange(10.6, 0.8)
numbers = [next(g) for _ in range(10)]
for i, n in enumerate(numbers):
    assert math.isclose(n, 10.6 + 0.8*i)
print(numbers)
print("OK R6")
````

````python
# VARIANTES R6: cambia SOLO la linea de actualizacion
def grange(init, ratio):      # progresion geometrica (multiplica)
    while True:
        yield init
        init *= ratio

def fib():                     # fibonacci
    a, b = 0, 1
    while True:
        yield a
        a, b = b, a + b

g = fib()
print([next(g) for _ in range(8)])
````

# 📖 R7 — Reemplazar por el mas/menos frecuente

### 🔍 Como reconocerlo
"caracter menos frecuente", "valor mas comun", "reemplazar los espacios/guiones por...".

### 🧠 Plan mental
1. `serie.value_counts()` → Series con frecuencias, ordenada de mayor a menor.
2. `.idxmin()` = el VALOR menos frecuente. `.idxmax()` = el mas frecuente. (idx porque quieres la ETIQUETA, no la cuenta).
3. Reemplazar con mascara: `serie[serie == ' '] = valor`.
4. **TRAMPA AVANZADA:** si el caracter a reemplazar (espacio, guion) tambien cuenta en las frecuencias, ¿debe excluirse? Mira el expected: si el espacio pudiera ser el "menos frecuente" y reemplazarse por si mismo, filtra primero: `serie[serie != ' '].value_counts()`.

### 📋 Plantilla

````python
# PLANTILLA R7 (version exacta del demo: menos frecuente)
def clean_whitespace_replace(serie):
    less_freq = serie.value_counts().idxmin()
    serie = serie.copy()
    serie[serie == ' '] = less_freq
    return serie

serie = pd.Series(list('bba cabc faabba aacbbfe'))
expected = pd.Series(list('bbaecabcefaabbaeaacbbfe'))
output = clean_whitespace_replace(serie)
assert expected.equals(output)
print("OK R7")

# Variante MAS frecuente: cambia idxmin -> idxmax
````

# 📖 R8 — Agrupaciones (groupby)

### 🔍 Como reconocerlo
"por genero", "por categoria", "media/suma/maximo de cada grupo", "cuantos hay de cada".

### 🧠 Plan mental
1. `df.groupby('columna_categorica')` no muestra nada solo → SIEMPRE va seguido de agregacion.
2. Agregaciones: `.count()`, `.mean(numeric_only=True)`, `.sum()`, `.max()`, `.describe()`.
3. Solo algunas columnas: `df.groupby('genero')[['edad','puntos']].mean()`.
4. Varias agregaciones a la vez: `.aggregate(['min','mean','max'])`.
5. FILTRAR grupos enteros por condicion del grupo: `.filter(funcion_que_devuelve_bool)`.

### ⚠️ Trampas
- Columnas de texto rompen `.mean()` → agrega `numeric_only=True`.
- Si agrupas por columna con NaN, esas filas DESAPARECEN del resultado.
- "cuantos usuarios son mujeres" NO necesita groupby: `len(df[df['Gender']=='F'])`.

### 📋 Plantilla

````python
# PLANTILLA R8
frame = pd.DataFrame({"nombre": ["Marisa","Laura","Manuel","Carlos"],
                      "edad": [34,34,11,30],
                      "puntos": [98,12,98,np.nan],
                      "genero": ["F","F","M","M"]})

display(frame.groupby('genero').count())
display(frame.groupby('genero').mean(numeric_only=True))
display(frame.groupby('genero')[['edad','puntos']].aggregate(['min','mean','max']))

# filtrar grupos cuya edad media > 30
def media(x):
    return x["edad"].mean() > 30
display(frame.groupby('genero').filter(media))

# conteo simple SIN groupby
print("Mujeres:", len(frame[frame['genero'] == 'F']))
````

# 📖 R9 — Filtrado con mascaras booleanas

### 🔍 Como reconocerlo
"filas donde", "mayores que", "que cumplan", "seleccionar los que", combinaciones con "y"/"o".

### 🧠 Plan mental
1. Filtro simple: `df[df['col'] > 5]`
2. Filtro + seleccion de columnas: `df.loc[df['col'] > 5, ['col1','col2']]`
3. Combinar condiciones: `&` (y), `|` (o), `~` (no). CADA condicion entre parentesis.
4. Pertenencia a lista: `df[df['col'].isin([...])]`
5. Contar cuantos cumplen: `mask.values.sum()` o `len(df[cond])`

### ⚠️ Trampas
- NUNCA `and`/`or` de Python con Series → error. Solo `&`, `|`, `~`.
- Sin parentesis en cada condicion → error de precedencia.
- Modificar un slice sin `.copy()` → SettingWithCopyWarning. La forma segura de modificar filtrado: `df.loc[cond, 'col'] = valor`.

### 📋 Plantilla

````python
# PLANTILLA R9
character_data = pd.DataFrame({'Name': ['bundenth','theorin','barlok'],
                               'Strength': [10,12,19],
                               'Wisdom': [20,13,6]}, index=['a','b','c'])

# filtro simple + columnas especificas
display(character_data.loc[character_data['Strength'] > 11, ['Name','Strength']])

# combinado con OR
display(character_data.loc[(character_data['Strength'] > 15) | (character_data['Wisdom'] > 15)])

# contar cuantos valores cumplen
frame_num = pd.DataFrame(np.random.randint(6, size=(2,3)), columns=list('ABC'))
mask = frame_num >= 2
print("Valores >= 2:", mask.values.sum())

# modificar de forma SEGURA los que cumplen (sin warning)
character_data.loc[character_data['Wisdom'] < 10, 'Wisdom'] = 0
display(character_data)
````

# 📖 R10 — Correlaciones

### 🔍 Como reconocerlo
"correlacion", "relacion entre variables", "cuales estan mas relacionadas".

### 🧠 Plan mental
1. Entre DOS columnas: `df['a'].corr(df['b'])` → un numero entre -1 y 1.
2. TODAS contra todas: `df.corr(numeric_only=True)` → matriz.
3. Interpretacion: cerca de 1 = suben juntas; cerca de -1 = una sube la otra baja; cerca de 0 = sin relacion lineal.
4. "La mas correlacionada con cada una" → OJO: la diagonal siempre es 1.0 (consigo misma). Anulala primero.

### 📋 Plantilla

````python
# PLANTILLA R10
datos = pd.DataFrame(np.random.randn(50, 4), columns=list('ABCD'))
datos['E'] = datos['A'] * 2 + np.random.randn(50) * 0.1   # E correlacionada con A a proposito

# dos columnas
print("corr A-E:", datos['A'].corr(datos['E']))

# matriz completa
corr = datos.corr(numeric_only=True)
display(corr)

# la mas correlacionada con cada una (anulando la diagonal con copia segura)
vals = corr.to_numpy().copy()
np.fill_diagonal(vals, 0)
corr_sin_diag = pd.DataFrame(vals, index=corr.index, columns=corr.columns)
print(corr_sin_diag.idxmax())
````

# 📖 R12 — Funciones Python (los clasicos de la lista de ejercicios)

### 🔍 Como reconocerlos y resolver cada uno
| Enunciado | Solucion en UNA linea de idea |
|---|---|
| "cuadrados de una lista" | list comprehension: `[x**2 for x in lista]` |
| "cantidad indeterminada de numeros, sin objetos complejos" | `def f(*args): return sum(args)` |
| "string al reves" | slicing: `s[::-1]` |
| "lo mismo con lambda" | `lambda s: s[::-1]` |
| "numero dentro de un rango" | `minimo <= n <= maximo` |
| "5 primeros multiplos, validar entrada" | `if n <= 0: print(error); return []` + comprehension |
| "tiene duplicados True/False" | `len(lista) != len(set(lista))` |
| "palindromo" | `s == s[::-1]` |
| "factorial recursivo" | caso base `if n == 0: return 1` + `n * f(n-1)` |

### ⚠️ Trampas
- "*sin* recibir objeto complejo" = OBLIGATORIO `*args`, no una lista.
- "funcion lambda que..." = te piden literalmente asignar un lambda a una variable, no un def.
- Validacion de entrada: mensaje de error + devolver lista VACIA (no None) si asi lo dice.

### 📋 Todas las plantillas juntas (ejecutables)

````python
# PLANTILLA R12 - los 9 clasicos
def cuadrados(lista):
    return [x**2 for x in lista]

def suma_numeros(*numeros):
    return sum(numeros)

def invertir(s):
    return s[::-1]

invertir_lambda = lambda s: s[::-1]

def en_rango(n, minimo, maximo):
    return minimo <= n <= maximo

def primeros_multiplos(n):
    if n <= 0:
        print("Error: el numero debe ser positivo")
        return []
    return [n*i for i in range(1, 6)]

def tiene_duplicados(lista):
    return len(lista) != len(set(lista))

tiene_duplicados_lambda = lambda lista: len(lista) != len(set(lista))

def es_palindromo(s):
    return s == s[::-1]

def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n-1)

# tests rapidos
print(cuadrados([1,2,3]))
print(suma_numeros(1,2,3,4))
print(invertir('hola'), invertir_lambda('hola'))
print(en_rango(5,1,10), en_rango(15,1,10))
print(primeros_multiplos(3), primeros_multiplos(-2))
print(tiene_duplicados([1,2,2]), tiene_duplicados_lambda([1,2,3]))
print(es_palindromo('reconocer'), es_palindromo('python'))
print(factorial(10))
````

# 📖 R13 — Analisis de dataset real (patron MovieLens del curso)

### 🔍 Como reconocerlo
Te dan un archivo (.dat, .csv) o URL con datos reales y preguntas tipo: "hay algun usuario que no tenga reviews?", "cuantos tienen menos de N?", "cuantos son de tal categoria?", "filtra los invalidos".

### 🧠 Plan mental (el flujo COMPLETO que uso el profe en clase)
1. **Cargar** con los parametros correctos (mirar el archivo primero si puedes):
   - Sin cabecera → `header=None` + `names=[...]` (si lo olvidas, PIERDES la primera fila de datos)
   - Separador raro (`::`) → `sep='::'` + `engine='python'`
   - Columna indice → `index_col=0`
2. **Explorar**: `sample(5)`, `head()`, `describe()`, `info()`, `dtypes`
3. **Contar por condicion**: `len(df[df['col'] == valor])`
4. **Grupos que existen** vs total: `len(df.groupby('col'))` = cuantos valores distintos tienen filas
5. **Filtrar por TAMANO de grupo** (el patron estrella): `.groupby('col')['col'].filter(lambda x: len(x) < N).value_counts()`
6. **Limpiar invalidos**: marcar con NaN (`df.loc[cond, 'col'] = np.nan`) y luego `drop` de esos indices.

### ⚠️ Trampas
- Olvidar `header=None` → la primera fila real se come como cabecera.
- `SettingWithCopyWarning`: nunca modifiques un slice; usa `df.loc[cond, 'col'] = valor` o trabaja sobre `.copy()`.
- "usuarios SIN reviews" = total de usuarios MENOS usuarios que aparecen en ratings (dos datasets, resta de conteos).

### 📋 Plantilla (con datos sinteticos que imitan MovieLens)

````python
# PLANTILLA R13 - simulo users y ratings como en el ejercicio real del curso
np.random.seed(0)
users = pd.DataFrame({'UserID': range(1, 21),
                      'Gender': np.random.choice(['F','M'], 20),
                      'Age': np.random.choice([1, 18, 25, 35, 50], 20)}).set_index('UserID')

# ratings: solo 15 de los 20 usuarios tienen reviews, con cantidades variables
uids = np.random.choice(range(1, 16), size=200)
ratings = pd.DataFrame({'UserID': uids,
                        'MovieID': np.random.randint(1, 50, size=200),
                        'Rating': np.random.randint(1, 6, size=200)})

# --- 1) carga real seria asi (referencia): ---
# users = pd.read_csv(ruta, sep='::', index_col=0, header=None,
#                     names=['UserID','Gender','Age','Occupation','Zip-code'], engine='python')

# --- 2) explorar ---
display(users.head(3))
display(ratings.sample(3))

# --- 3) contar por condicion: cuantas mujeres ---
print("Mujeres:", len(users[users['Gender'] == 'F']))

# --- 4) usuarios SIN ninguna review (patron resta de conteos) ---
n_con_reviews = len(ratings.groupby('UserID'))
n_total = len(users)
print(f"Con reviews: {n_con_reviews} | Total: {n_total} | SIN reviews: {n_total - n_con_reviews}")

# --- 5) cuantos tienen menos de 10 reviews (patron filter por tamano de grupo) ---
menos_de_10 = ratings.groupby('UserID')['UserID'].filter(lambda x: len(x) < 10).value_counts()
display(menos_de_10)
print("Usuarios con menos de 10 reviews:", len(menos_de_10))

# --- 6) limpiar edades invalidas (Age==1): marcar NaN y eliminar ---
users.loc[users['Age'] == 1, 'Age'] = np.nan
users.drop(users[pd.isnull(users['Age'])].index, inplace=True)
print("Usuarios tras limpieza:", len(users))
````

### 🔄 Variantes
- "guardar el resultado limpio" → `df.to_csv(ruta, sep=',', na_rep='null')`
- "media de rating por usuario" → `ratings.groupby('UserID')['Rating'].mean()`
- "el usuario con MAS reviews" → `ratings['UserID'].value_counts().idxmax()`
- "describir por grupo" → `users.groupby('Gender').describe()`
