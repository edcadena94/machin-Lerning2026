## 01
- Dado un diccionario con una lista de ciudades y su poblacion concatenada, crear una funcion que devuelva un DataFrame con dos columnas:
  - Columna 1, el nombre de la ciudad sin la poblacion
  - Columna 2, la poblacion de la ciudad

### 🧭 Guia paso a paso
1. **Identificar:** "concatenada" + "separar en columnas" → tabla de decision → **Receta R1** (parsear strings).
2. **Contrato:** mira la celda del profe abajo → funcion se llama `extract_population`, devuelve DataFrame, la poblacion va entre comillas → **texto, no numero**.
3. **Plan:** convertir dict a DataFrame → crear columna nueva con `apply(lambda x: x.split('-')[1].strip())` → sobrescribir la original con `[0]`.
4. **Trampa:** el orden — primero extrae la poblacion, DESPUES sobrescribe el nombre (si no, el guion desaparece antes de tiempo). Y strip() SIEMPRE.

```python
# Your solution HERE !!!!
def extract_population(ciudades):
    df = pd.DataFrame(ciudades)                                        # dict -> tabla
    df['Population'] = df['City'].apply(lambda x: x.split('-')[1].strip())  # pedazo derecho, limpio
    df['City'] = df['City'].apply(lambda x: x.split('-')[0].strip())        # pedazo izquierdo, limpio
    return df
```

```python
ciudades = {'City': ["Quito - 2011388", "Guayaquil - 2650288",
                     "Cuenca - 636996", "Santo Domingo - 458580"]}
expected_output = pd.DataFrame({'City': ["Quito", "Guayaquil", "Cuenca", "Santo Domingo"],
                                'Population': ["2011388", "2650288", "636996", "458580"]})
output = extract_population(ciudades)   # This will call your function extract_population
display(output)
assert(expected_output.equals(output))  # This will fail if the result is not as expected
print("EJERCICIO 01: APROBADO ✅")
```

## 02
- Crear una funcion que reciba una lista y devuelva True si tiene elementos duplicados y False si no los tiene.

### 🧭 Guia paso a paso
1. **Identificar:** "duplicados" + "True/False" (no pide marcar ni eliminar, solo SI/NO) → **Receta R12**, truco del set.
2. **Contrato:** funcion `has_duplicates`, devuelve un booleano.
3. **Plan:** un `set` elimina repetidos automaticamente. Si la lista original tiene 4 elementos y el set queda con 3, es porque habia un repetido. Comparo tamanos con `len()`.
4. **La logica:** `len(lista) != len(set(lista))` → si los tamanos son DISTINTOS devuelve True (habia duplicados).

```python
# Your solution HERE !!!!
def has_duplicates(lista):
    return len(lista) != len(set(lista))   # tamanos distintos = habia repetidos
```

```python
lista1 = [3, 7, 1, 9]
lista2 = [3, 7, 3, 9]
print(has_duplicates(lista1))
print(has_duplicates(lista2))
assert(has_duplicates(lista1) == False)
assert(has_duplicates(lista2) == True)
print("EJERCICIO 02: APROBADO ✅")
```

## 03
- Crear una funcion que reciba un array de NumPy y devuelva otro en el cual esten marcados como True los elementos duplicados (a partir de la segunda ocurrencia), y como False los no repetidos o las primeras ocurrencias.

### 🧭 Guia paso a paso
1. **Identificar:** "marcados como True los duplicados a partir de la segunda ocurrencia" → **Receta R2-A**: esa frase describe EXACTAMENTE lo que hace `.duplicated()`.
2. **Contrato:** funcion `find_duplicates`, el test compara con `np.array_equal` → debe devolver un **array**, no una Series → agrega `.values` al final.
3. **Plan:** convertir el array a `pd.Series` (porque duplicated es de pandas) → llamar `.duplicated()` → `.values` para volver a array.
4. **Trampa:** si devuelves la Series sin `.values`, `np.array_equal` puede funcionar igual, pero es mas seguro devolver el array puro.

```python
# Your solution HERE !!!!
def find_duplicates(numbers):
    return pd.Series(numbers).duplicated().values   # Series para usar duplicated, .values para volver a array
```

```python
numbers = np.array([5, 5, 2, 8, 2, 9, 5, 1])
expected_output = np.array([False, True, False, False, True, False, True, False])
output = find_duplicates(numbers)   # This will call your function find_duplicates
print(output)
assert(np.array_equal(expected_output, output))
print("EJERCICIO 03: APROBADO ✅")
```

## 05
- Resolucion de dos sistemas de ecuaciones:
  - A: 4x + 3y = 18 ; -2y + 5x = 1
  - B: 2x + y + z = 9 ; x - y + 2z = 5 ; 3x + 2y - z = 4
- Verifica por codigo los resultados

### 🧭 Guia paso a paso
1. **Identificar:** "sistema de ecuaciones" + "verifica" → **Receta R3**.
2. **Contrato:** no hay funcion con nombre — es codigo directo. Pero SI exige verificar.
3. **Plan:** matriz A = coeficientes (una fila por ecuacion, x primero, y despues, z al final) → vector b = resultados → `linalg.solve(A, b)` → verificar cada ecuacion con `math.isclose`.
4. **TRAMPA GRANDE:** la segunda ecuacion de A viene DESORDENADA: "-2y + 5x = 1". Reordenala mentalmente a "5x - 2y = 1" → la fila es [5, -2], NO [-2, 5].
5. **Recordatorio:** coeficiente invisible = 1 ("x" solo = 1x), "menos la letra" = -1.

```python
# Your solution HERE !!!!
import numpy.linalg as linalg

# Sistema A: 4x + 3y = 18 ; 5x - 2y = 1  (la segunda venia desordenada, la reordene)
a = np.array([[4, 3],
              [5, -2]])
b = np.array([18, 1])
x, y = linalg.solve(a, b)
print(f'x={x} y={y}')
assert math.isclose(4*x + 3*y, 18)
assert math.isclose(5*x - 2*y, 1)

# Sistema B: 2x + y + z = 9 ; x - y + 2z = 5 ; 3x + 2y - z = 4
A2 = np.array([[2, 1, 1],
               [1, -1, 2],
               [3, 2, -1]])
b2 = np.array([9, 5, 4])
x, y, z = linalg.solve(A2, b2)
print(f'x={x} y={y} z={z}')
assert math.isclose(2*x + y + z, 9)
assert math.isclose(x - y + 2*z, 5)
assert math.isclose(3*x + 2*y - z, 4)
print("EJERCICIO 05: APROBADO ✅")
```

## 08
- Escribir un generador `countdown` que cada vez que se invoque devuelva el siguiente numero de una cuenta regresiva a partir de un valor inicial, restando un step en cada paso
- Debe soportar valores numericos en coma flotante
- La funcion potencialmente se puede llamar un numero infinito de veces
- Usando plain Python, sin ninguna libreria

### 🧭 Guia paso a paso
1. **Identificar:** "generador" + "cada vez que se invoque" + "infinito" + "sin ninguna libreria" → **Receta R6** (yield). Las 4 palabras magicas juntas.
2. **Contrato:** funcion `countdown(init, step)`, se consume con `next(g)`. PROHIBIDO numpy (dice "sin ninguna libreria").
3. **Plan:** la estructura FIJA de R6: `while True:` + `yield valor` + actualizar. Aqui la actualizacion es RESTAR: `init -= step`.
4. **Trampa:** sin el `while True`, el generador se agota al primer next y el segundo da StopIteration. "Infinitas veces" = while True obligatorio.

```python
# Your solution HERE !!!!
def countdown(init, step):
    while True:          # dispensador infinito
        yield init       # entrega el valor actual y SE PAUSA aqui
        init -= step     # al proximo next: resta el paso y repite
```

```python
init = 20.5
step = 1.5
g = countdown(init, step)   # This calls your function countdown

iters = 8
numbers = [next(g) for n in range(iters)]
for i, n in enumerate(numbers):
    assert(math.isclose(n, init - (step * i)))   # This will fail if the countdown is wrong
print(numbers)
print("EJERCICIO 08: APROBADO ✅")
```

## 09
- Crear una funcion que reciba una Serie de Pandas y devuelva esa misma serie, reemplazando los asteriscos por el caracter mas frecuente en dicha serie.

### 🧭 Guia paso a paso
1. **Identificar:** "caracter mas frecuente" + "reemplazando" → **Receta R7** (value_counts + idx).
2. **Contrato:** funcion `clean_asterisk_replace`, devuelve la Series. Aqui es el MAS frecuente → `.idxmax()` (el demo usaba el menos → idxmin; solo cambia esa palabra).
3. **Plan:** contar frecuencias con `value_counts()` → sacar la etiqueta ganadora con `idxmax()` → reemplazar con mascara `serie[serie == '*'] = ganador`.
4. **Trampa fina:** ¿el asterisco cuenta en las frecuencias? Aqui hay pocos asteriscos asi que no gana, pero la forma BLINDADA es excluirlo antes de contar: `serie[serie != '*'].value_counts().idxmax()`. Usa la blindada siempre.
5. **Buena practica:** `.copy()` para no modificar la serie original del test.

```python
# Your solution HERE !!!!
def clean_asterisk_replace(serie):
    mas_frecuente = serie[serie != '*'].value_counts().idxmax()  # excluyo * antes de contar (blindado)
    serie = serie.copy()                                          # copia para no tocar la original
    serie[serie == '*'] = mas_frecuente                           # donde hay *, pongo el ganador
    return serie
```

```python
serie = pd.Series(list('aab*cabc*aabba*aacbb'))
expected_output = pd.Series(list('aabacabcaaabbaaaacbb'))
output = clean_asterisk_replace(serie)  # This calls your function clean_asterisk_replace
assert(expected_output.equals(output))  # This will fail if result is not as expected
print(''.join(output.values))
print("EJERCICIO 09: APROBADO ✅")
```

## 10
- Dado un DataFrame de ventas con columnas Vendedor y Monto:
  - Calcular el monto total vendido por cada vendedor
  - Averiguar cuantas ventas hizo cada vendedor
  - Devolver el nombre del vendedor con mas ventas (mayor numero de operaciones)

### 🧭 Guia paso a paso
1. **Identificar:** "por cada vendedor" (por cada = por grupo) → **Receta R8** (groupby). "el que MAS" → idxmax de R7.
2. **Contrato:** funcion `analizar_ventas(df)` que devuelve una tupla de 3 cosas: (Series de totales, Series de conteos, nombre del top).
3. **Plan:**
   - Total por grupo: `df.groupby('Vendedor')['Monto'].sum()`
   - Conteo por grupo: `df['Vendedor'].value_counts()` (mas directo que groupby+count)
   - El de mas ventas: `.idxmax()` sobre el conteo → devuelve el NOMBRE (idx = la etiqueta, no el numero).
4. **Trampa:** "mas ventas" = mas OPERACIONES (conteo), no mas dinero (suma). Lee dos veces que te piden.

```python
# Your solution HERE !!!!
def analizar_ventas(df):
    totales = df.groupby('Vendedor')['Monto'].sum()   # separa por vendedor y suma el monto de cada monton
    conteos = df['Vendedor'].value_counts()           # cuantas filas (ventas) tiene cada vendedor
    top = conteos.idxmax()                            # la ETIQUETA (nombre) del que mas se repite
    return totales, conteos, top
```

```python
ventas = pd.DataFrame({'Vendedor': ['Ana','Luis','Ana','Marta','Luis','Ana'],
                       'Monto':    [100,  250,   80,   300,    120,   50]})

totales, conteos, top = analizar_ventas(ventas)   # This calls your function analizar_ventas
print(totales)
print(conteos)
print("Top vendedor:", top)

assert(totales['Ana'] == 230)          # 100 + 80 + 50
assert(totales['Luis'] == 370)         # 250 + 120
assert(conteos['Ana'] == 3)
assert(top == 'Ana')                   # Ana tiene 3 ventas, mas que nadie
print("EJERCICIO 10: APROBADO ✅")
```
