```python
productos = {'Item': ["Laptop Gamer - 1200", "Mouse Inalambrico - 25",
                      "Teclado Mecanico - 80", "Monitor 4K, Curvo - 450"]}
expected_output = pd.DataFrame({'Item': ["Laptop Gamer", "Mouse Inalambrico",
                                         "Teclado Mecanico", "Monitor 4K, Curvo"],
                                'Price': ["1200", "25", "80", "450"]})
output = extract_price(productos)   # This will call your function extract_price
display(output)
assert(expected_output.equals(output))  # This will fail if the result is not as expected
```

```python
# Your solution HERE !!!!
def extract_price(productos):
    df = pd.DataFrame(productos)
    df['Price'] = df['Item'].apply(lambda x: x.split('-')[1].strip())
    df['Item'] = df['Item'].apply(lambda x: x.split('-')[0].strip())
    return df
```

```python
def repeated_element(array_src):
    vistos = set()
    for elemento in array_src:
        if elemento in vistos:
            return elemento  # Encontró el repetido y lo devuelve
        vistos.add(elemento)
```
